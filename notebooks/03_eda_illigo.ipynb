## 1. Data Loading and Schema Analysis

```python
# List all JSON files
data_dir = Path('../data/raw/illigo')
json_files = list(data_dir.glob('*.json'))

print(f"Found {len(json_files)} JSON event log files")

# Load sample file to examine structure
if json_files:
    with open(json_files[0], 'r') as f:
        sample_events = json.load(f)
    
    print(f"\nSample event structure from {json_files[0].name}:")
    if isinstance(sample_events, list) and len(sample_events) > 0:
        print(json.dumps(sample_events[0], indent=2))
```
